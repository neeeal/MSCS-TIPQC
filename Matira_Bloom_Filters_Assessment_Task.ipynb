# Bloom Filters Assessment Task


##  Function Initialization

```python
## Static Vars
m = 32
k = 3
bloomFilter = [0 for x in range(m)]

## Hashing Algorithms
# Algo 1: Simple Additive (Weakest, prone to anagram collisions)
def hash_ascii_1(element, size):
    return sum([ord(char) for char in element]) % size

# Algo 2: Positional Weighting (Fixes anagrams by weighing char by index)
def hash_ascii_2(element, size):
    val = 0
    for i, char in enumerate(element):
        val += ord(char) * (i + 1)
    return val % size

# Algo 3: Rolling Hash (Stronger distribution using a prime multiplier)
def hash_ascii_3(element, size):
    val = 0
    prime = 31
    for char in element:
        val = (val * prime + ord(char))
    return val % size

# List of hash functions for iteration
hash_functions = [hash_ascii_1, hash_ascii_2, hash_ascii_3]

## Insert Function
def insert_bloom(bloomFilter, element):
    # Iterate through the specific function list, not just a range number
    for h_func in hash_functions:
        position = h_func(element, m)
        bloomFilter[position] = 1
        print(f"Inserted '{element}' at position {position} using {h_func.__name__}")

## Check/Verify Function (Required to actually use a Bloom Filter)
def check_bloom(bloomFilter, element):
    for h_func in hash_functions:
        position = h_func(element, m)
        if bloomFilter[position] == 0:
            return False # Definitely not in set
    return True # Probably in set
```

## Test Code

```python
## Test Twice
insert_bloom(bloomFilter, "Data")
insert_bloom(bloomFilter, "Science")

print(f"\nFinal Filter: {bloomFilter}")
print(f"Check 'Data': {check_bloom(bloomFilter, 'Data')}")
print(f"Check 'Python': {check_bloom(bloomFilter, 'Python')}")
```

## Simulation Set

```python
# 1. Define Inputs (10 Items)
inputs = [
    "kafka", "spark", "hadoop", "python", "java",
    "scala", "docker", "airflow", "redis", "mongo"
]

# 2. Insert Phase
print("--- INSERT PHASE ---")
for item in inputs:
    insert_bloom(bloomFilter, item)

print(f"\nFinal Bit Array ({sum(bloomFilter)}/32 bits set):")
print(bloomFilter)

# 3. Check Phase (10 Checks)
# 5 items that ARE in the set
checks_exist = ["kafka", "spark", "java", "redis", "mongo"]
# 5 items that are NOT in the set
checks_not_exist = ["rust", "golang", "ruby", "perl", "swift"]

print("\n--- CHECK PHASE ---")
print(f"{'QUERY':<10} | {'EXPECTED':<10} | {'RESULT':<10}")
print("-" * 35)

# Check Existing items
for item in checks_exist:
    result = check_bloom(bloomFilter, item)
    print(f"{item:<10} | {'True':<10} | {str(result):<10}")

# Check Non-Existing items
for item in checks_not_exist:
    result = check_bloom(bloomFilter, item)
    print(f"{item:<10} | {'False':<10} | {str(result):<10}")
```
