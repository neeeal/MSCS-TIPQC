## Data Cleaning using `fillna()`

```python
sample_df = df.copy()
```

```python
sample_df.isna().sum()
```

```python
sample_df.fillna(method="ffill", inplace=True)
```

```python
sample_df.isna().sum()
```

## Data Cleaning using `dropduplicates()`

```python
sample_df = df.copy()
```

```python
key_dupes = sample_df[sample_df.duplicated(subset=['Sales_Order #'], keep=False)]
key_dupes
```

```python
sample_df = sample_df.drop_duplicates(subset=['Sales_Order #'], keep='first')
```

```python
key_dupes = sample_df[sample_df.duplicated(subset=['Sales_Order #'], keep=False)]
key_dupes
```
