## Treap Class Initialization

```python
# @title
import random

class TreapNode:
    def __init__(self, key, priority):
        self.key = key
        self.priority = priority
        self.left = None
        self.right = None
        self.parent = None  # Explicit parent link as per text [cite: 170]

class Treap:
    def __init__(self):
        self.root = None

    def _right_rotate(self, x):
        """
        Performs a right rotation promoting node x over its parent y.
        Adapted from Listing 3.2[cite: 237].
        """
        y = x.parent
        p = y.parent

        # 1. Link y's left child to x's right child
        if x.right:
            x.right.parent = y
        y.left = x.right

        # 2. Link x's right child to y
        x.right = y
        y.parent = x

        # 3. Link p to x
        x.parent = p
        if p is None:
            self.root = x
        elif p.left == y:
            p.left = x
        else:
            p.right = x

    def _left_rotate(self, x):
        """
        Performs a left rotation promoting node x over its parent y.
        Adapted from Listing 3.3[cite: 338].
        """
        y = x.parent
        p = y.parent

        # 1. Link y's right child to x's left child
        if x.left:
            x.left.parent = y
        y.right = x.left

        # 2. Link x's left child to y
        x.left = y
        y.parent = x

        # 3. Link p to x
        x.parent = p
        if p is None:
            self.root = x
        elif p.left == y:
            p.left = x
        else:
            p.right = x

    def insert(self, key, priority):
        """
        Inserts a new node and bubbles it up to restore Max-Heap property.
        Adapted from Listing 3.5[cite: 537].
        """
        # --- Step 1: Standard BST Insertion ---
        new_node = TreapNode(key, priority)
        if self.root is None:
            self.root = new_node
            return

        parent = None
        current = self.root
        while current:
            parent = current
            if key < current.key:
                current = current.left
            else:
                current = current.right

        # Place new node
        new_node.parent = parent
        if key < parent.key:
            parent.left = new_node
        else:
            parent.right = new_node

        # --- Step 2: Bubble Up (Fix Heap Property) ---
        # User Requirement: Max-Heap (Child Priority > Parent Priority violates heap)
        while new_node.parent and new_node.priority > new_node.parent.priority:
            parent = new_node.parent
            if new_node == parent.left:
                self._right_rotate(new_node) # [cite: 589]
            else:
                self._left_rotate(new_node)  # [cite: 591]

    def print_tree(self, node, level=0, prefix="Root: "):
        """Helper to visualize the tree structure."""
        if node is not None:
            print(" " * (level * 4) + prefix + f"[{node.key} | P:{node.priority}]")
            if node.left or node.right:
                if node.left:
                    self.print_tree(node.left, level + 1, "L--- ")
                else:
                    print(" " * ((level + 1) * 4) + "L--- None")
                if node.right:
                    self.print_tree(node.right, level + 1, "R--- ")
                else:
                    print(" " * ((level + 1) * 4) + "R--- None")
```

## Task A: The Average Case (Randomized)

```python
print("--- TASK A: Average Case (Randomized) ---")
treap_a = Treap()
keys_a = [10, 20, 30, 40, 50]
print(f"Inserting keys: {keys_a} with random priorities...")
for k in keys_a:
    p = random.randint(1, 100) # Random priority
    treap_a.insert(k, p)
treap_a.print_tree(treap_a.root)
print("\n")
```

## Task B: The Best Case (Perfectly Balanced)

```python
print("--- TASK B: Best Case (Perfectly Balanced) ---")
treap_b = Treap()
data_b = [
    (40, 100),
    (20, 80), (60, 80),
    (10, 60), (30, 60), (50, 60), (70, 60)
]
random.shuffle(data_b)
print(f"Inserting data (shuffled input): {data_b}")
for k, p in data_b:
    treap_b.insert(k, p)
treap_b.print_tree(treap_b.root)
print("\n")
```

## Task C: The "Worst Case" (Degenerate Path)

```python
print("--- TASK C: Worst Case (Degenerate Path) ---")
treap_c = Treap()
keys_c = [10, 20, 30, 40, 50]
priorities_c = [50, 40, 30, 20, 10] # Strictly decreasing
print(f"Inserting keys: {keys_c}")
print(f"With priorities: {priorities_c}")

for i in range(len(keys_c)):
    treap_c.insert(keys_c[i], priorities_c[i])

treap_c.print_tree(treap_c.root)
```
